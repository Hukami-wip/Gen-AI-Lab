# Experiment: Generative Temporal State Model (GTSM)

Conventional generative sequence models are typically trained to predict only the immediate next element of the sequence from previous tokens.

This experiment explores a new task and training procedure, which aims to teach the model a more flexible, continuous representation of sequential data. The core idea is to train the model to predict elements at arbitrary temporal locations, using a dynamic context of both past and future elements.

The design used in this experiment would enable a model to operate in three distinct modes of inference:

1. **Next-Element Prediction:** 
    
    Given a history of past elements, the model can function like a standard autoregressive model, predicting the element at the next canonical position.
2. **Goal Generation:**
    
    Given a history, the model can be asked to generate a plausible, distant future element, effectively setting a long-term objective.
3. **Path Interpolation:**
    
    Given a history and a future "goal" element, the model can be tasked with generating an intermediate element, filling in the steps needed to connect the present to the desired future.
